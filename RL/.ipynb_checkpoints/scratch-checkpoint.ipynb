```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline

from data_handling import read_data, process_timestamp, get_min_since, transforms
```

```python
plt.rcParams['figure.figsize'] = [20, 10]
```

```python
data = read_data('MSFT')
min_since = get_min_since(data)
data_df = process_timestamp(data, min_since)
```

```python
t = transforms(data_df)
t.all_transforms()
```

```python
data_df = t.transformed
```

```python
fig, ax = plt.subplots()
for company in ['MSFT', 'AAPL', 'AMZN', 'FB', 'GOOG', 'ROKU', 'SNAP', 'TSLA', 'TWTR', 'UBER']:
    data = read_data(company)
    min_since = get_min_since(data)
    data_df = process_timestamp(data, min_since)
    t = transforms(data_df)
    t.all_transforms()
    data_df = t.transformed
    
    ax.plot(data_df['close'], label = company )


ax.legend()
plt.show()
```
